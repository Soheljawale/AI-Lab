```python
print(wt)
```

```python
from nltk.probability import FreqDist
```

```python
fd = FreqDist(s)
```

```python
print(fd)
```

```python
fc = FreqDist(tt.lower() for tt in wt)
```

```python
fc
```

```python
fc.most_common(3)
```

```python
fc.plot()
```
